# **Manejo de datos faltantes:**

- Se decidió imputar los datos faltantes de la variable satisfacción utilizando su promedio, ya que es un método directo y efectivo para tratar variables cuantitativas sin perder registros ni alterar la tendencia central de la muestra.


- Se decidió eliminar las filas que contenían datos faltantes, ya que representaban menos del 5 % del total de los registros. Por ello, se consideró que esta era la solución más adecuada para tratar los valores ausentes sin comprometer la integridad de la muestra.

    1. weight_kg --> No es adecuado imputar peso con una media (o mediana) simple porque el peso depende de la especie, raza, tamaño y edad

    2. household_size --> Depende del tipo de vivienda y ubicación del adoptante; imputar un promedio o mediana distorsionaría el entorno real.

    3. microchipped --> Siendo binaria, rellenarla sería asumir un estado sin evidencia.

# **Creación de variables**

```python
df['annual_food_cost_usd']=df['monthly_food_cost_usd']*12
cost_columns=['annual_food_cost_usd','annual_vet_cost_usd',
              'annual_grooming_cost_usd',
              'annual_toys_accessories_cost_usd',
              'annual_training_cost_usd',
              'annual_insurance_cost_usd']

df['total_annual_cost_usd']=df[cost_columns].sum(axis=1)
df['gasto_medico_total_usd']=df['annual_vet_cost_usd']+df['annual_insurance_cost_usd']
df['gasto_extra_anual_usd']=df['annual_grooming_cost_usd']+df['annual_toys_accessories_cost_usd']+df['annual_training_cost_usd']
df['pct_gasto_comida']=((df['annual_food_cost_usd']/df['total_annual_cost_usd'])*100).round(2)
nuevas_variables=['gasto_medico_total_usd','gasto_extra_anual_usd','pct_gasto_comida']

display(df[['species']+nuevas_variables].head())
display(df.groupby('species')[nuevas_variables].mean().round(2))
```

#  **Visualizaciones:**

```python
import matplotlib.pyplot as plt

df['annual_food_cost_usd'] = df['monthly_food_cost_usd'] * 12

cost_columns = [
    'annual_food_cost_usd',
    'annual_vet_cost_usd',
    'annual_grooming_cost_usd',
    'annual_toys_accessories_cost_usd',
    'annual_training_cost_usd',
    'annual_insurance_cost_usd',
]
cost_labels = [
    'Comida',
    'Veterinaria',
    'Peluquería / Higiene',
    'Juguetes y Accesorios',
    'Entrenamiento',
    'Seguro Médico',
]

gastos_promedio = df.groupby('species')[cost_columns].mean()
gastos_promedio.columns = cost_labels

colores = plt.cm.Set2.colors[: len(cost_labels)]

fig, axes = plt.subplots(nrows=2, ncols=4, figsize=(16, 8))
axes_flat = axes.flatten()

formato_pct = lambda p: f'{p:.0f}%' if p >= 4 else ''

for i, especie in enumerate(gastos_promedio.index):
  ax = axes_flat[i]
  valores = gastos_promedio.loc[especie]

  wedges, _, _ = ax.pie(
      valores,
      autopct=formato_pct,
      startangle=140,
      colors=colores,
      pctdistance=0.75,
      wedgeprops=dict(width=0.85, edgecolor='white', linewidth=1.5),
  )
  ax.set_title(especie, fontsize=13, fontweight='bold', pad=5)

ax_legend = axes_flat[7]
ax_legend.axis('off')
ax_legend.legend(
    wedges,
    cost_labels,
    title='Rubros de Gasto',
    title_fontsize=12,
    fontsize=11,
    loc='center',
    frameon=True,
    edgecolor='gray',
)

plt.suptitle(
    'Distribución del Gasto Anual por Especie',
    fontsize=16,
    fontweight='bold',
    y=0.98,
)
plt.tight_layout()

# Esta línea guarda la gráfica en alta resolución
plt.savefig('distribucion_gasto_anual.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns

seguro_por_especie = (
    df.groupby("species")["has_pet_insurance"].mean() * 100
).sort_values(ascending=False)
promedio_general = df["has_pet_insurance"].mean() * 100

fig, ax = plt.subplots(figsize=(10, 5.5))

# Paleta Set2 ajustada a la cantidad de especies
palette_seguro = sns.color_palette("Set2", n_colors=len(seguro_por_especie))

# Gráfico de barras
bars = ax.bar(
    seguro_por_especie.index,
    seguro_por_especie.values,
    color=palette_seguro,
    edgecolor="black",
    linewidth=0.8,
    width=0.55,
)

# Agregar etiquetas de porcentaje sobre cada barra
for bar in bars:
  altura = bar.get_height()
  ax.annotate(
      f"{altura:.1f}%",
      xy=(bar.get_x() + bar.get_width() / 2, altura),
      xytext=(0, 4),
      textcoords="offset points",
      ha="center",
      va="bottom",
      fontsize=10.5,
      fontweight="bold",
  )

# Línea punteada que muestra el promedio general de la población
ax.axhline(
    promedio_general,
    color="#e74c3c",
    linestyle="--",
    linewidth=1.2,
    label=f"Promedio General ({promedio_general:.1f}%)",
)

# Títulos y estética
plt.title(
    "Porcentaje de Mascotas con Seguro Médico por Especie",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Especie", fontsize=11, fontweight="bold", labelpad=10)
plt.ylabel(
    "Tasa de Aseguramiento (%)", fontsize=11, fontweight="bold", labelpad=10
)
plt.ylim(0, max(seguro_por_especie.values) + 8)

# Formato de porcentaje en el eje Y
ax.yaxis.set_major_formatter(ticker.PercentFormatter(decimals=0))

plt.xticks(fontsize=10.5)
plt.yticks(fontsize=10)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.legend(loc="upper right", frameon=True, fontsize=10)
sns.despine(top=True, right=True)

plt.tight_layout()

# Guarda la gráfica en alta calidad para subirla a GitHub
plt.savefig("tasa_aseguramiento_especie.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns

# Preparar los datos agrupados por mes
adopciones_mes = df["adoption_month"].value_counts().sort_index()
promedio_mensual = adopciones_mes.mean()

meses_es = [
    "Ene", "Feb", "Mar", "Abr", "May", "Jun",
    "Jul", "Ago", "Sep", "Oct", "Nov", "Dic"
]

fig, ax = plt.subplots(figsize=(10.5, 5.5))

# Colores de la paleta Set2 para las franjas estacionales
palette_set2 = sns.color_palette("Set2")
color_primavera = palette_set2[2]
color_navidad = palette_set2[1]

# Franjas verticales para resaltar los picos estacionales
ax.axvspan(
    3.5, 6.5, color=color_primavera, alpha=0.22,
    label="Temporada Primavera / Inicio Verano (Abr–Jun)"
)
ax.axvspan(
    10.5, 12.3, color=color_navidad, alpha=0.22,
    label="Temporada Festiva / Fin de Año (Nov–Dic)"
)

# Gráfico de línea con marcadores y área suave debajo
ax.plot(
    adopciones_mes.index,
    adopciones_mes.values,
    color="#2b8cbe",
    marker="o",
    markersize=8,
    linewidth=2.5,
    markeredgecolor="black",
    markeredgewidth=0.8,
    zorder=4,
    label="Adopciones por Mes",
)
ax.fill_between(
    adopciones_mes.index, adopciones_mes.values, color="#2b8cbe", alpha=0.12
)

# Línea punteada que muestra el promedio mensual general
ax.axhline(
    promedio_mensual,
    color="#e74c3c",
    linestyle="--",
    linewidth=1.2,
    zorder=3,
    label=f"Promedio Mensual ({promedio_mensual:,.0f} adopciones)".replace(",", "."),
)

# Agregar etiquetas numéricas sobre/debajo de cada punto
for mes, cantidad in zip(adopciones_mes.index, adopciones_mes.values):
    if mes in [2, 3, 7, 9, 10]:
        offset_y = -15
        va = "top"
    else:
        offset_y = 9
        va = "bottom"

    ax.annotate(
        f"{cantidad:,}".replace(",", "."),
        xy=(mes, cantidad),
        xytext=(0, offset_y),
        textcoords="offset points",
        ha="center",
        va=va,
        fontsize=9.5,
        fontweight="bold",
        color="#222222",
    )

# Títulos y estética
plt.title(
    "Adopción de Mascotas por Mes",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Mes de Adopción", fontsize=11, fontweight="bold", labelpad=10)
plt.ylabel("Cantidad de Adopciones", fontsize=11, fontweight="bold", labelpad=10)

plt.xticks(ticks=range(1, 13), labels=meses_es, fontsize=10.5)
plt.yticks(fontsize=10)
plt.xlim(0.7, 12.3)
plt.ylim(700, 2000)

ax.yaxis.set_major_formatter(
    ticker.FuncFormatter(lambda x, pos: f"{int(x):,}".replace(",", "."))
)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.legend(loc="upper left", frameon=True, fontsize=9.5)
sns.despine(top=True, right=True)

plt.tight_layout()

# Guarda la gráfica en alta calidad para subirla a GitHub
plt.savefig("estacionalidad_adopcion_mascotas.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import pandas as pd
import numpy as np
import seaborn as sns

df["tramo_edad"] = pd.cut(
    df["age_years"],
    bins=[0, 2, 5, 8, 12, 20],
    labels=[
        "0–2 años\n(Cachorro/Joven)",
        "2–5 años\n(Adulto Joven)",
        "5–8 años\n(Adulto)",
        "8–12 años\n(Senior)",
        "12+ años\n(Geriátrico)",
    ],
)

tabla_cronicas = pd.crosstab(df["tramo_edad"], df["chronic_condition"], normalize="index") * 100
tabla_cronicas.columns = ["Sin Condición Crónica", "Con Condición Crónica"]

fig, ax = plt.subplots(figsize=(10, 5.2))
x_pos = np.arange(len(tabla_cronicas))

ax.stackplot(
    x_pos,
    tabla_cronicas["Con Condición Crónica"],
    tabla_cronicas["Sin Condición Crónica"],
    labels=["Con Condición Crónica (%)", "Sin Condición Crónica (%)"],
    colors=["#fc8d62", "#66c2a5"],
    alpha=0.85,
)

# Línea de frontera y etiquetas exactas del riesgo
ax.plot(
    x_pos,
    tabla_cronicas["Con Condición Crónica"],
    color="#d95f02",
    marker="o",
    markersize=8,
    linewidth=2.5,
    markeredgecolor="black",
)
for idx, pct in enumerate(tabla_cronicas["Con Condición Crónica"]):
    ax.annotate(
        f"{pct:.1f}%",
        xy=(idx, pct),
        xytext=(0, 10),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=10.5,
        fontweight="bold",
        color="#222222",
        bbox=dict(boxstyle="round,pad=0.25", facecolor="white", edgecolor="#d95f02", alpha=0.9),
    )

plt.title("Riesgo de Enfermedad Crónica según la Etapa de Vida de la Mascota", fontsize=13, fontweight="bold", pad=15)
plt.xlabel("Etapa de Vida (Edad en Años)", fontsize=11, fontweight="bold", labelpad=10)
plt.ylabel("Proporción de Mascotas (%)", fontsize=11, fontweight="bold", labelpad=10)
plt.xticks(x_pos, tabla_cronicas.index, fontsize=10, fontweight="bold")
plt.xlim(0, len(tabla_cronicas) - 1)
plt.ylim(0, 100)
ax.yaxis.set_major_formatter(ticker.PercentFormatter(decimals=0))
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.legend(loc="upper left", frameon=True, fontsize=10)
sns.despine(top=True, right=True)

plt.tight_layout()
plt.savefig("area_cronicas_por_edad.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

care_cols = ["vaccinated", "spayed_neutered", "microchipped", "has_pet_insurance"]
for c in care_cols:
    df[c] = df[c].astype(bool)

care_summary = (df.groupby("species")[care_cols].mean() * 100).round(1)

especies_radar = ["Dog", "Cat", "Rabbit", "Bird"]
nombres_radar = {
    "Dog": "Perro (Dog)",
    "Cat": "Gato (Cat)",
    "Rabbit": "Conejo (Rabbit)",
    "Bird": "Ave (Bird)",
}
categorias_radar = [
    "Vacunación\n(vaccinated)",
    "Esterilización\n(spayed_neutered)",
    "Microchip\n(microchipped)",
    "Seguro Médico\n(has_pet_insurance)",
]

angles = np.linspace(0, 2 * np.pi, len(categorias_radar), endpoint=False).tolist()
angles += angles[:1]

fig, ax = plt.subplots(figsize=(7.5, 7.5), subplot_kw=dict(polar=True))
palette_radar = sns.color_palette("Set2", n_colors=4)

for idx, esp in enumerate(especies_radar):
    valores = care_summary.loc[esp].tolist()
    valores += valores[:1]
    ax.plot(
        angles, valores, linewidth=2.5, marker="o", markersize=7,
        color=palette_radar[idx], label=nombres_radar[esp]
    )
    ax.fill(angles, valores, color=palette_radar[idx], alpha=0.18)

ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categorias_radar, fontsize=10.5, fontweight="bold")
ax.set_ylim(0, 100)
ax.set_yticks([20, 40, 60, 80, 100])
ax.set_yticklabels(["20%", "40%", "60%", "80%", "100%"], fontsize=8.5, color="#555555")
ax.grid(True, linestyle="--", alpha=0.5)

plt.title("Perfil de Tenencia Responsable y Prevención por Especie", fontsize=13, fontweight="bold", pad=22)
plt.legend(loc="upper right", bbox_to_anchor=(1.25, 1.12), frameon=True, fontsize=10)

plt.tight_layout()
plt.savefig("radar_tenencia_responsable.png", dpi=300, bbox_inches="tight")
plt.show()
```
